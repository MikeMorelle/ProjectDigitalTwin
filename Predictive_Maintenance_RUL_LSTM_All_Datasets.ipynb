```python
import numpy as np

def create_sequences(df, seq_length, feature_columns):

    X = []
    y = []

    engine_ids = df['unit_number'].unique()

    for engine_id in engine_ids:

        engine_data = df[df['unit_number'] == engine_id]

        feature_array = engine_data[feature_columns].values
        rul_array = engine_data['RUL'].values

        for i in range(len(engine_data) - seq_length):

            sequence = feature_array[i:i + seq_length]

            target = rul_array[i + seq_length]

            X.append(sequence)
            y.append(target)

    return np.array(X), np.array(y)
```

```python
X_train, y_train = create_sequences(
    train_split,
    SEQ_LENGTH,
    feature_columns
)
print(X_train.shape)
print(X_train[0])
print(y_train.shape)
```

```python
X_val, y_val = create_sequences(
    val_split,
    SEQ_LENGTH,
    feature_columns
)
print(X_val.shape)
print(y_val.shape)
```

```python
class LSTMModel(nn.Module):

    def __init__(self, input_size):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=64,
            num_layers=2,
            batch_first=True,
            dropout=0.2
        )

        self.fc = nn.Sequential(
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 1)
        )

    def forward(self, x):

        out, (hidden, cell) = self.lstm(x)

        out = out[:, -1, :]

        out = self.fc(out)

        return out.squeeze()
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = LSTMModel(input_size=len(feature_columns)).to(device)
```

```python
criterion = nn.L1Loss()  # MAE
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
EPOCHS = 20

for epoch in range(EPOCHS):

    model.train()

    total_loss = 0

    for X_batch, y_batch in train_loader:

        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        optimizer.zero_grad()

        predictions = model(X_batch)

        loss = criterion(predictions, y_batch)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)

    print(f"Epoch {epoch+1}, Loss: {avg_loss:.4f}")
```

```python
#####################################################################
# INSTALLS
#####################################################################

!pip install wandb

#####################################################################
# IMPORTS
#####################################################################

import os
import wandb
import torch
import torch.nn as nn

from torch.utils.data import DataLoader

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np

#####################################################################
# GOOGLE DRIVE SAVE PATHS
#####################################################################

from google.colab import drive
drive.mount('/content/drive')

# Main project directory
PROJECT_DIR = "/content/drive/MyDrive/Predictive_Maintenance"
Training_Name = "/lstm_fd001_seq"

# Create folders automatically
os.makedirs(PROJECT_DIR, exist_ok=True)
os.makedirs(f"{PROJECT_DIR}{Training_Name}/checkpoints", exist_ok=True)
os.makedirs(f"{PROJECT_DIR}{Training_Name}/best_model", exist_ok=True)

#####################################################################
# WANDB LOGIN
#####################################################################

# wandb authentication
wandb.login()

#####################################################################
# INITIALIZE WANDB PROJECT
#####################################################################

wandb.init(
    project="NASA-CMAPSS-LSTM-RUL",
    name="lstm_fd001_baseline",

    config={
        "architecture": "LSTM",
        "dataset": "NASA CMAPSS FD001",
        "sequence_length": SEQ_LENGTH,
        "hidden_size": 64,
        "num_layers": 2,
        "dropout": 0.2,
        "batch_size": 32,
        "learning_rate": 0.001,
        "epochs": 30,
        "loss_function": "MAE"
    }
)

config = wandb.config

#####################################################################
# TENSORS
#####################################################################

X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)

X_val = torch.tensor(X_val, dtype=torch.float32)
y_val = torch.tensor(y_val, dtype=torch.float32)

#####################################################################
# DATASETS + DATALOADERS
#####################################################################

train_dataset = torch.utils.data.TensorDataset(X_train, y_train)
val_dataset = torch.utils.data.TensorDataset(X_val, y_val)

train_loader = DataLoader(
    train_dataset,
    batch_size=config.batch_size,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=config.batch_size
)

#####################################################################
# MODEL
#####################################################################

class LSTMModel(nn.Module):

    def __init__(self, input_size):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=config.hidden_size,
            num_layers=config.num_layers,
            batch_first=True,
            dropout=config.dropout
        )

        self.fc = nn.Sequential(
            nn.Linear(config.hidden_size, 32),
            nn.ReLU(),
            nn.Linear(32, 1)
        )

    def forward(self, x):

        out, (hidden, cell) = self.lstm(x)

        # last timestep output
        out = out[:, -1, :]

        out = self.fc(out)

        return out.squeeze()

#####################################################################
# DEVICE
#####################################################################

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using device:", device)

#####################################################################
# MODEL INIT
#####################################################################

model = LSTMModel(
    input_size=len(feature_columns)
).to(device)

#####################################################################
# LOSS + OPTIMIZER
#####################################################################

criterion = nn.L1Loss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=config.learning_rate
)

#####################################################################
# OPTIONAL: WATCH MODEL
#####################################################################

wandb.watch(model, log="all", log_freq=100)

#####################################################################
# TRAINING LOOP
#####################################################################

best_val_loss = float("inf")

for epoch in range(config.epochs):

    #############################################################
    # TRAINING
    #############################################################

    model.train()

    train_loss = 0

    for X_batch, y_batch in train_loader:

        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        optimizer.zero_grad()

        predictions = model(X_batch)

        loss = criterion(predictions, y_batch)

        loss.backward()

        optimizer.step()

        train_loss += loss.item()

    avg_train_loss = train_loss / len(train_loader)

    #############################################################
    # VALIDATION
    #############################################################

    model.eval()

    val_loss = 0

    all_predictions = []
    all_targets = []

    with torch.no_grad():

        for X_batch, y_batch in val_loader:

            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            predictions = model(X_batch)

            loss = criterion(predictions, y_batch)

            val_loss += loss.item()

            #####################################################
            # STORE PREDICTIONS FOR METRICS
            #####################################################

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_targets.extend(
                y_batch.cpu().numpy()
            )
    avg_val_loss = val_loss / len(val_loader)

    #############################################################
    # REGRESSION METRICS
    #############################################################

    mae = mean_absolute_error(
        all_targets,
        all_predictions
    )

    mse = mean_squared_error(
        all_targets,
        all_predictions
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        all_targets,
        all_predictions
    )

    #############################################################
    # PRINT
    #############################################################

    print(
        f"Epoch [{epoch+1}/{config.epochs}] | "
        f"Train Loss: {avg_train_loss:.4f} | "
        f"Val Loss: {avg_val_loss:.4f}"
    )

    #############################################################
    # WANDB LOGGING
    #############################################################

    wandb.log({

    "epoch": epoch + 1,

    #########################################################
    # LOSSES
    #########################################################

    "train_loss": avg_train_loss,
    "val_loss": avg_val_loss,

    #########################################################
    # REGRESSION METRICS
    #########################################################

    "mae": mae,
    "mse": mse,
    "rmse": rmse,
    "r2_score": r2,

    #########################################################
    # LEARNING RATE
    #########################################################

    "learning_rate": optimizer.param_groups[0]["lr"],


    "prediction_scatter":
        wandb.plot.scatter(
            table=wandb.Table(
                data=[
                    [x, y]
                    for x, y in zip(
                        all_targets,
                        all_predictions
                    )
                ],
                columns=["True", "Predicted"]
            ),
            x="True",
            y="Predicted",
            title="Prediction Scatter"
        )
    })

    #############################################################
    # SAVE CHECKPOINT EACH EPOCH
    #############################################################

    checkpoint_path = (
        f"{PROJECT_DIR}{Training_Name}/checkpoints/"
        f"lstm_epoch_{epoch+1}.pth"
    )

    torch.save({
        "epoch": epoch + 1,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "train_loss": avg_train_loss,
        "val_loss": avg_val_loss,
    }, checkpoint_path)

    #############################################################
    # SAVE BEST MODEL
    #############################################################

    if avg_val_loss < best_val_loss:

        best_val_loss = avg_val_loss

        best_model_path = (
            f"{PROJECT_DIR}{Training_Name}/best_model/"
            f"best_lstm_model.pth"
        )

        torch.save({
            "epoch": epoch + 1,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "val_loss": avg_val_loss,
        }, best_model_path)

        print("Best model saved!")

#####################################################################
# SAVE FINAL MODEL
#####################################################################

FINAL_MODEL_PATH = (
    f"{PROJECT_DIR}{Training_Name}/final_lstm_model.pth"
)

torch.save(model.state_dict(), FINAL_MODEL_PATH)

print(f"Final model saved to:\n{FINAL_MODEL_PATH}")

#####################################################################
# FINISH WANDB
#####################################################################

wandb.finish()
```
